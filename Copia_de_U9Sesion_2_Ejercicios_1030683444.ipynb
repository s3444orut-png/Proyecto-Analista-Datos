1 EJERCICIO

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

db_tb = pd.read_csv('https://drive.google.com/uc?export=download&id=1JS7jkpbWMaOlpsh_7Q6P8nue1U9aiM-C')
db_tb['Total Vendidas']=db_tb['Cantidad Vendida']*db_tb['Precio Unitario'].astype(float)
db_tb['Precio Unitario']=db_tb['Precio Unitario'].astype(float)
db_tb['Cantidad Vendida']=db_tb['Cantidad Vendida'].astype(int)
print(db_tb.head())

sns.boxplot(x=db_tb['Categoría'],y=db_tb['Total Vendidas'])
plt.show()


```

2 EJERCICIO

```python

db_tb.head()

map_cal = db_tb.select_dtypes(include=np.number)
Valr_2 = map_cal.corr()

plt.figure(figsize=(12,8))
sns.heatmap(Valr_2, annot=True, cmap='coolwarm', fmt=".2f")
plt.title("Map_Cal")
plt.show()
```

3 EJERCICIO

```python
print(db_tb.head())

plt.figure(figsize=(14,8))
#Grafico de Velas = boxplot
sns.boxplot(x=db_tb['Producto'],y=db_tb['Total Vendidas'], data= db_tb)
plt.title('Valor de Ventas por Productos')
plt.xlabel('Productos')
plt.xlabel('Valor')
plt.show()
```

4 EJERCICIO

```python
db_tb

tb1 = pd.pivot_table(db_tb, values = 'Total Vendidas', index='Categoría',columns='Producto', aggfunc='sum')
tb1

plt.figure(figsize=(12,8))
sns.heatmap(tb1, annot=True, cmap='coolwarm', fmt=".2f", linewidths=.5)
plt.title('Ventas por distubucion de procustos')
plt.xlabel('Productos')
plt.ylabel('Valor Total')
plt.show()

```
