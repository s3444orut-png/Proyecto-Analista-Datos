1 EJERCICIO

```python
import seaborn as sns
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df= pd.read_csv('https://drive.google.com/uc?export=download&id=1VbrQjDF2Br0tjHHiruUnp-0qmnQwon1I')

#Grafico de disperción scatterplot
sns.scatterplot(data = df, x='Edad', y='Ingreso Mensual')
#Grafico de linea de Tendencia regplot
sns.regplot(data=df,x='Edad', y='Ingreso Mensual')

```

2 EJERCICIO

```python
df
G_R = df.groupby('Region')['Gasto Mensual'].mean()
G_R_S = df.groupby('Region')['Gasto Mensual'].sum()
G_R_=pd.DataFrame(G_R).T
print(G_R.head())
G_R_S=pd.DataFrame(G_R_S)

G_R_.plot(kind='bar',stacked=True)
plt.xticks(rotation=0)
plt.title('Gastos por Región')
plt.show()


# bar se usa para grupos de data
sns.barplot(x='Region', y='Gasto Mensual',data=G_R_S, color='green')
plt.title('Gasto Mensual por región')
plt.xlabel('Region')
plt.xlabel('Valor_G')
plt.show()

#G_2 = df.groupby(['Edad','Region'])
G_E =df.groupby(['Edad','Region'])['Gasto Mensual'].sum().reset_index()
G_E=G_E.pivot(index='Edad',columns='Region', values='Gasto Mensual')
G_E = pd.DataFrame(G_E)
plt.figure(figsize=(20,26))
G_E.plot(kind='bar',stacked=True)
plt.xticks(rotation=0,fontsize=10)
plt.title('Gastos por Edad y Región')
plt.show()
```

```python
df

for z in range(len(df)):
 if df.loc[z,'Edad']<=17:
  df.loc[z,'Segmento']= 'Menor de Edad'
 elif df.loc[z,'Edad']>17 and df.loc[z,'Edad']<31 :
   df.loc[z,'Segmento']= 'Jovenes'
 elif df.loc[z,'Edad']>30 and df.loc[z,'Edad']<60 :
   df.loc[z,'Segmento']= 'Adultos'
 else:
   df.loc[z,'Segmento']= 'Adulto Mayor'

sns.kdeplot(data=df, x='Ingreso Mensual', hue='Segmento')

df_Groupby = df.groupby('Segmento')['Ingreso Mensual'].sum()

print(df_Groupby.head())







```
